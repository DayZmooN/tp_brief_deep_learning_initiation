Exercice 1 le perceptron

```python
import numpy as np
```

```python
X = np.array([[0, 0], [0, 1], [1,0],[1,1]])
y = np.array([0,0,0,1])

np.random.seed(42)
w = np.random.randn(2)
b = np.random.randn()
```

```python

def sigmoid(z):
    return 1 / (1+ np.exp(-z))
```

```python
def predict(X, w, b):
    return sigmoid(X@ w + b)
```

```python
def bce(y,p,eps=1e-12):
    return -np.mean(y * np.log(p+ eps) + (1 - y) * np.log(1 - p + eps))
```

```python
lr, epochs = 0.5, 2000
for i in range(epochs):
    p = predict(X,w, b)
    dz = p- y
    w -= lr * (X.T @ dz) / len(X)
    b -= lr * dz.mean()
    if i % 200 == 0:
        print(f"Epoch {i}, Loss: { bce(y,p):.4f}")
        
print("Prédiction :", predict(X,w,b).round(3))
```
